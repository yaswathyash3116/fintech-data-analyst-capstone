```python
import os

import pandas as pd
import matplotlib.pyplot as plt

sip = pd.read_csv("../data/04_monthly_sip_inflows_cleaned.csv")

plt.figure(figsize=(12,6))
plt.plot(sip['month'], sip['yoy_growth_pct'], marker='o')

plt.title("YoY SIP Growth Percentage")
plt.xlabel("Month")
plt.ylabel("Growth %")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Insight

YoY SIP growth remained relatively low in the initial period, increased significantly during the middle of the timeline, reaching a peak above 50%, and later showed a gradual decline. This indicates strong growth in SIP participation followed by a period of stabilization.

```python
import pandas as pd
import matplotlib.pyplot as plt

funds = pd.read_csv("../data/01_fund_master_cleaned.csv")

category_counts = funds['category'].value_counts()

plt.figure(figsize=(10,6))
category_counts.plot(kind='bar')

plt.title("Fund Category Distribution")
plt.xlabel("Category")
plt.ylabel("Number of Funds")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Insight

The distribution of mutual funds across categories highlights which investment segments have the highest number of schemes. Categories with more funds indicate greater product diversity and investor choice.

```python
import pandas as pd
import matplotlib.pyplot as plt

funds = pd.read_csv("../data/01_fund_master_cleaned.csv")

top_fund_houses = funds['fund_house'].value_counts().head(10)

plt.figure(figsize=(10,6))
top_fund_houses.plot(kind='bar')

plt.title("Top 10 Fund Houses by Number of Schemes")
plt.xlabel("Fund House")
plt.ylabel("Number of Schemes")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Insight

The chart highlights the fund houses with the highest number of schemes in the dataset. Fund houses with more schemes generally offer a wider variety of investment options across different categories and risk profiles.

```python
import pandas as pd
import matplotlib.pyplot as plt

funds = pd.read_csv("../data/01_fund_master_cleaned.csv")

risk_counts = funds['risk_category'].value_counts()

plt.figure(figsize=(10,6))
risk_counts.plot(kind='bar')

plt.title("Risk Category Distribution")
plt.xlabel("Risk Category")
plt.ylabel("Number of Funds")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Insight

The distribution of funds across risk categories shows the risk appetite targeted by mutual fund schemes. A higher concentration in specific risk categories indicates greater product availability for those investor profiles.
