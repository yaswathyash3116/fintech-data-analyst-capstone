```python
import pandas as pd

funds = pd.read_csv("../data/01_fund_master_cleaned.csv")
nav = pd.read_csv("../data/02_nav_history_cleaned.csv")
sip = pd.read_csv("../data/04_monthly_sip_inflows_cleaned.csv")
```

```python
print(funds.shape)
print(nav.shape)
print(sip.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(sip['month'], sip['sip_inflow_crore'])
plt.title("Monthly SIP Inflows Trend")
plt.xlabel("Month")
plt.ylabel("SIP Inflow (Crore)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Insight 1
Monthly SIP inflows have grown steadily over the observed period, indicating increasing investor confidence and higher participation in mutual fund SIP investments.
